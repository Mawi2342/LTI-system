## Task 1a, creating two signals with different frequencies

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
import sounddevice as sd

# Generate a time vector ’t’ and the incrementing ∆t = 0,01ms as suggested in the assignment
dt = 0.0001
t = np.arange(0, 5, dt)

## For x1: period T=0.01s, amplitude A=1, phase shift φ = 0 and f= 1/T = 1/0,01 = 100
## For x2: amplitude A=1, phase shift φ = -pi/4 
x1 = 1*np.sin(2*np.pi*100*t+0)
x2 = 1*np.sin(2*np.pi*1000*t-np.pi/4)

# Create a new figure
fig, ax = plt.subplots()

ax.plot(t, x1, label='x1')
ax.plot(t, x2, label='x2')
ax.set_xlabel('Time s')
ax.set_ylabel('Amplitude')
ax.set_xlim(0, 0.01)
ax.set_ylim(-1.5 , 1.5)

ax.legend()
ax.grid()

plt.show()

#sd.play(x1 ,1/dt ,blocking=True)
sd.play(x2 ,1/dt ,blocking=True)
```
